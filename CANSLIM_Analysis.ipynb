```python
# @title
# CANSLIM Stock Analysis System
# A comprehensive implementation of William O'Neil's CANSLIM methodology

# Install required packages
!pip install yfinance pandas numpy matplotlib seaborn plotly requests beautifulsoup4 ta-lib

import yfinance as yf
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.graph_objects as go
from plotly.subplots import make_subplots
import requests
from bs4 import BeautifulSoup
import warnings
warnings.filterwarnings('ignore')

# Set up plotting style
plt.style.use('seaborn-v0_8')
sns.set_palette("husl")

class CANSLIMAnalyzer:
    def __init__(self, symbol, period='2y'):
        """
        Initialize CANSLIM analyzer for a given stock symbol

        Parameters:
        symbol (str): Stock ticker symbol
        period (str): Time period for analysis ('1y', '2y', '5y', etc.)
        """
        self.symbol = symbol.upper()
        self.period = period
        self.stock = yf.Ticker(self.symbol)

        # Fetch data
        self.data = self.stock.history(period=period)
        self.info = self.stock.info
        self.financials = self.stock.financials
        self.quarterly_financials = self.stock.quarterly_financials

        # CANSLIM scores
        self.canslim_scores = {}

    def analyze_current_earnings(self):
        """
        C - Current Quarterly Earnings
        Look for 25%+ earnings growth in recent quarters
        """
        try:
            # Get quarterly earnings data
            quarterly_earnings = self.stock.quarterly_earnings

            if quarterly_earnings is None or quarterly_earnings.empty:
                return 0, "No quarterly earnings data available"

            # Calculate earnings growth rates
            earnings_growth = []
            for i in range(1, min(5, len(quarterly_earnings))):
                current = quarterly_earnings.iloc[i-1]['Earnings']
                previous = quarterly_earnings.iloc[i]['Earnings']

                if previous != 0:
                    growth = ((current - previous) / abs(previous)) * 100
                    earnings_growth.append(growth)

            if not earnings_growth:
                return 0, "Insufficient earnings data for growth calculation"

            # Score based on recent earnings growth
            recent_growth = np.mean(earnings_growth[:2])  # Last 2 quarters

            if recent_growth >= 25:
                score = 100
            elif recent_growth >= 15:
                score = 80
            elif recent_growth >= 10:
                score = 60
            elif recent_growth >= 5:
                score = 40
            elif recent_growth >= 0:
                score = 20
            else:
                score = 0

            analysis = f"Recent earnings growth: {recent_growth:.1f}%"
            return score, analysis

        except Exception as e:
            return 0, f"Error analyzing current earnings: {str(e)}"

    def analyze_annual_earnings(self):
        """
        A - Annual Earnings Growth
        Look for 25%+ annual earnings growth over 3-5 years
        """
        try:
            # Get annual earnings data
            if self.financials is None or self.financials.empty:
                return 0, "No annual financial data available"

            # Get net income (earnings) data
            if 'Net Income' not in self.financials.index:
                return 0, "Net income data not available"

            net_income = self.financials.loc['Net Income'].dropna()

            if len(net_income) < 2:
                return 0, "Insufficient annual earnings data"

            # Calculate year-over-year growth rates
            growth_rates = []
            for i in range(1, len(net_income)):
                current = net_income.iloc[i-1]
                previous = net_income.iloc[i]

                if previous != 0:
                    growth = ((current - previous) / abs(previous)) * 100
                    growth_rates.append(growth)

            if not growth_rates:
                return 0, "Cannot calculate annual growth rates"

            # Score based on average annual growth
            avg_growth = np.mean(growth_rates)

            if avg_growth >= 25:
                score = 100
            elif avg_growth >= 20:
                score = 80
            elif avg_growth >= 15:
                score = 60
            elif avg_growth >= 10:
                score = 40
            elif avg_growth >= 0:
                score = 20
            else:
                score = 0

            analysis = f"Average annual earnings growth: {avg_growth:.1f}%"
            return score, analysis

        except Exception as e:
            return 0, f"Error analyzing annual earnings: {str(e)}"

    def analyze_new_products(self):
        """
        N - New Products, Services, or Management
        This is harder to quantify, so we'll use revenue growth as a proxy
        """
        try:
            if 'Total Revenue' not in self.financials.index:
                return 50, "Revenue data not available - using neutral score"

            revenue = self.financials.loc['Total Revenue'].dropna()

            if len(revenue) < 2:
                return 50, "Insufficient revenue data"

            # Calculate revenue growth
            recent_revenue = revenue.iloc[0]
            previous_revenue = revenue.iloc[1]

            if previous_revenue != 0:
                revenue_growth = ((recent_revenue - previous_revenue) / abs(previous_revenue)) * 100
            else:
                return 50, "Cannot calculate revenue growth"

            # Score based on revenue growth (proxy for new products/services)
            if revenue_growth >= 20:
                score = 100
            elif revenue_growth >= 15:
                score = 80
            elif revenue_growth >= 10:
                score = 60
            elif revenue_growth >= 5:
                score = 40
            elif revenue_growth >= 0:
                score = 20
            else:
                score = 0

            analysis = f"Revenue growth: {revenue_growth:.1f}% (proxy for new products/services)"
            return score, analysis

        except Exception as e:
            return 50, f"Error analyzing new products: {str(e)}"

    def analyze_supply_demand(self):
        """
        S - Supply and Demand (Volume Analysis)
        Look for increasing volume on up days and decreasing volume on down days
        """
        try:
            # Calculate price changes and volume patterns
            self.data['Price_Change'] = self.data['Close'].pct_change()
            self.data['Volume_MA'] = self.data['Volume'].rolling(window=20).mean()
            self.data['Volume_Ratio'] = self.data['Volume'] / self.data['Volume_MA']

            # Analyze recent volume patterns (last 20 days)
            recent_data = self.data.tail(20)

            up_days = recent_data[recent_data['Price_Change'] > 0]
            down_days = recent_data[recent_data['Price_Change'] < 0]

            if len(up_days) > 0 and len(down_days) > 0:
                avg_volume_up = up_days['Volume_Ratio'].mean()
                avg_volume_down = down_days['Volume_Ratio'].mean()

                volume_pattern = avg_volume_up - avg_volume_down

                if volume_pattern > 0.3:
                    score = 100
                elif volume_pattern > 0.1:
                    score = 80
                elif volume_pattern > 0:
                    score = 60
                elif volume_pattern > -0.1:
                    score = 40
                else:
                    score = 20

                analysis = f"Volume pattern score: {volume_pattern:.2f} (higher volume on up days is positive)"
            else:
                score = 50
                analysis = "Insufficient price movement data for volume analysis"

            return score, analysis

        except Exception as e:
            return 50, f"Error analyzing supply and demand: {str(e)}"

    def analyze_leader_laggard(self):
        """
        L - Leader or Laggard
        Compare stock performance to market (S&P 500)
        """
        try:
            # Get S&P 500 data for comparison
            spy = yf.download('^GSPC', period=self.period)['Close']

            # Calculate relative performance
            stock_return = (self.data['Close'].iloc[-1] / self.data['Close'].iloc[0] - 1) * 100
            market_return = (spy.iloc[-1] / spy.iloc[0] - 1) * 100

            relative_performance = stock_return - market_return

            # Score based on relative performance
            if relative_performance > 20:
                score = 100
            elif relative_performance > 10:
                score = 80
            elif relative_performance > 0:
                score = 60
            elif relative_performance > -10:
                score = 40
            elif relative_performance > -20:
                score = 20
            else:
                score = 0

            analysis = f"Stock return: {stock_return:.1f}%, Market return: {market_return:.1f}%, Relative: {relative_performance:.1f}%"
            return score, analysis

        except Exception as e:
            return 50, f"Error analyzing leader/laggard: {str(e)}"

    def analyze_institutional_sponsorship(self):
        """
        I - Institutional Sponsorship
        Use trading volume as a proxy for institutional interest
        """
        try:
            # Calculate average volume metrics
            avg_volume = self.data['Volume'].mean()
            recent_volume = self.data['Volume'].tail(20).mean()

            # Get market cap for context
            market_cap = self.info.get('marketCap', 0)

            # Score based on volume and market cap (proxies for institutional interest)
            volume_increase = (recent_volume / avg_volume - 1) * 100

            if market_cap > 10e9 and volume_increase > 20:  # Large cap with volume increase
                score = 100
            elif market_cap > 2e9 and volume_increase > 10:  # Mid cap with volume increase
                score = 80
            elif market_cap > 500e6 and volume_increase > 0:  # Small cap with volume increase
                score = 60
            elif market_cap > 100e6:  # Micro cap
                score = 40
            else:
                score = 20

            analysis = f"Market cap: ${market_cap/1e9:.1f}B, Volume change: {volume_increase:.1f}%"
            return score, analysis

        except Exception as e:
            return 50, f"Error analyzing institutional sponsorship: {str(e)}"

    def analyze_market_direction(self):
        """
        M - Market Direction
        Analyze overall market trend using S&P 500
        """
        try:
            # Get S&P 500 data
            spy = yf.download('^GSPC', period='6mo')['Close']

            # Calculate moving averages
            spy_ma20 = spy.rolling(window=20).mean()
            spy_ma50 = spy.rolling(window=50).mean()

            current_price = spy.iloc[-1]
            ma20 = spy_ma20.iloc[-1]
            ma50 = spy_ma50.iloc[-1]

            # Score based on market trend
            if current_price > ma20 > ma50:
                score = 100
                trend = "Strong uptrend"
            elif current_price > ma20:
                score = 80
                trend = "Uptrend"
            elif current_price > ma50:
                score = 60
                trend = "Neutral to positive"
            elif current_price < ma20 < ma50:
                score = 20
                trend = "Downtrend"
            else:
                score = 40
                trend = "Mixed signals"

            analysis = f"Market trend: {trend}"
            return score, analysis

        except Exception as e:
            return 50, f"Error analyzing market direction: {str(e)}"

    def calculate_canslim_score(self):
        """
        Calculate overall CANSLIM score and individual component scores
        """
        print(f"Analyzing {self.symbol} using CANSLIM methodology...")
        print("=" * 50)

        # Analyze each component
        c_score, c_analysis = self.analyze_current_earnings()
        a_score, a_analysis = self.analyze_annual_earnings()
        n_score, n_analysis = self.analyze_new_products()
        s_score, s_analysis = self.analyze_supply_demand()
        l_score, l_analysis = self.analyze_leader_laggard()
        i_score, i_analysis = self.analyze_institutional_sponsorship()
        m_score, m_analysis = self.analyze_market_direction()

        # Store scores
        self.canslim_scores = {
            'Current Earnings (C)': c_score,
            'Annual Earnings (A)': a_score,
            'New Products (N)': n_score,
            'Supply & Demand (S)': s_score,
            'Leader/Laggard (L)': l_score,
            'Institutional (I)': i_score,
            'Market Direction (M)': m_score
        }

        # Calculate overall score
        overall_score = np.mean(list(self.canslim_scores.values()))

        # Print results
        print(f"C - Current Earnings: {c_score}/100 - {c_analysis}")
        print(f"A - Annual Earnings: {a_score}/100 - {a_analysis}")
        print(f"N - New Products: {n_score}/100 - {n_analysis}")
        print(f"S - Supply & Demand: {s_score}/100 - {s_analysis}")
        print(f"L - Leader/Laggard: {l_score}/100 - {l_analysis}")
        print(f"I - Institutional: {i_score}/100 - {i_analysis}")
        print(f"M - Market Direction: {m_score}/100 - {m_analysis}")
        print("=" * 50)
        print(f"Overall CANSLIM Score: {overall_score:.1f}/100")

        return overall_score

    def generate_signals(self):
        """
        Generate buy/sell signals based on CANSLIM analysis
        """
        overall_score = self.calculate_canslim_score()

        # Technical indicators for timing
        self.data['SMA_20'] = self.data['Close'].rolling(window=20).mean()
        self.data['SMA_50'] = self.data['Close'].rolling(window=50).mean()

        current_price = self.data['Close'].iloc[-1]
        sma_20 = self.data['SMA_20'].iloc[-1]
        sma_50 = self.data['SMA_50'].iloc[-1]

        # Generate signals
        if overall_score >= 70 and current_price > sma_20 > sma_50:
            signal = "STRONG BUY"
            color = "green"
        elif overall_score >= 60 and current_price > sma_20:
            signal = "BUY"
            color = "lightgreen"
        elif overall_score >= 40 and current_price > sma_50:
            signal = "HOLD"
            color = "yellow"
        elif overall_score < 40 and current_price < sma_20:
            signal = "SELL"
            color = "orange"
        else:
            signal = "STRONG SELL"
            color = "red"

        print(f"\n🎯 SIGNAL: {signal}")
        print(f"Current Price: ${current_price:.2f}")
        print(f"20-day SMA: ${sma_20:.2f}")
        print(f"50-day SMA: ${sma_50:.2f}")

        return signal, overall_score

    def plot_analysis(self):
        """
        Create comprehensive visualization of the analysis
        """
        fig = make_subplots(
            rows=3, cols=2,
            subplot_titles=('Price Chart with Moving Averages', 'CANSLIM Component Scores',
                          'Volume Analysis', 'Price vs Volume',
                          'Relative Performance vs S&P 500', 'Technical Indicators'),
            specs=[[{"secondary_y": True}, {"type": "bar"}],
                   [{"secondary_y": True}, {"secondary_y": True}],
                   [{"secondary_y": False}, {"secondary_y": False}]]
        )

        # Price chart with moving averages
        fig.add_trace(
            go.Scatter(x=self.data.index, y=self.data['Close'], name='Close Price'),
            row=1, col=1
        )
        fig.add_trace(
            go.Scatter(x=self.data.index, y=self.data['SMA_20'], name='20-day SMA'),
            row=1, col=1
        )
        fig.add_trace(
            go.Scatter(x=self.data.index, y=self.data['SMA_50'], name='50-day SMA'),
            row=1, col=1
        )

        # CANSLIM component scores
        components = list(self.canslim_scores.keys())
        scores = list(self.canslim_scores.values())

        fig.add_trace(
            go.Bar(x=components, y=scores, name='CANSLIM Scores'),
            row=1, col=2
        )

        # Volume analysis
        fig.add_trace(
            go.Scatter(x=self.data.index, y=self.data['Volume'], name='Volume'),
            row=2, col=1
        )
        fig.add_trace(
            go.Scatter(x=self.data.index, y=self.data['Volume_MA'], name='Volume MA'),
            row=2, col=1, secondary_y=True
        )

        # Price vs Volume scatter
        fig.add_trace(
            go.Scatter(x=self.data['Volume'], y=self.data['Close'],
                      mode='markers', name='Price vs Volume'),
            row=2, col=2
        )

        # Relative performance (simplified)
        returns = self.data['Close'].pct_change().cumsum() * 100
        fig.add_trace(
            go.Scatter(x=self.data.index, y=returns, name='Cumulative Returns'),
            row=3, col=1
        )

        # Technical indicators summary
        recent_data = self.data.tail(20)
        fig.add_trace(
            go.Scatter(x=recent_data.index, y=recent_data['Close'],
                      name='Recent Price Action'),
            row=3, col=2
        )

        fig.update_layout(height=1000, showlegend=True,
                         title_text=f"CANSLIM Analysis for {self.symbol}")
        fig.show()

        # CANSLIM scores bar chart
        plt.figure(figsize=(12, 6))
        bars = plt.bar(components, scores, color=['red' if s < 50 else 'yellow' if s < 70 else 'green' for s in scores])
        plt.title(f'CANSLIM Component Scores for {self.symbol}')
        plt.ylabel('Score (0-100)')
        plt.xticks(rotation=45)
        plt.axhline(y=70, color='green', linestyle='--', alpha=0.7, label='Strong Buy Threshold')
        plt.axhline(y=50, color='orange', linestyle='--', alpha=0.7, label='Neutral Threshold')
        plt.legend()
        plt.tight_layout()
        plt.show()

# Example usage and analysis
def analyze_multiple_stocks(symbols, period='1y'):
    """
    Analyze multiple stocks using CANSLIM methodology
    """
    results = {}

    for symbol in symbols:
        print(f"\n{'='*60}")
        print(f"ANALYZING {symbol}")
        print(f"{'='*60}")

        try:
            analyzer = CANSLIMAnalyzer(symbol, period)
            signal, score = analyzer.generate_signals()
            results[symbol] = {
                'signal': signal,
                'score': score,
                'analyzer': analyzer
            }

            # Create visualization
            analyzer.plot_analysis()

        except Exception as e:
            print(f"Error analyzing {symbol}: {str(e)}")
            results[symbol] = {
                'signal': 'ERROR',
                'score': 0,
                'analyzer': None
            }

    return results

# Run analysis on sample stocks
# sample_stocks = ['OUST', 'APP', 'NVAX', 'OLO', 'NVDA', 'USFD', 'CRS', 'PFGC', 'CLS', 'EVLV']
sample_stocks = ['APP','TOST','NVAX','OLO','EVLV','XERS','LIF','BDSX','DCTH','ISSC','CURI','MPWR','PLMR','BDSX','DCTH']
# sample_stocks = ['CRS','RYAAY','LMT','RTX','NOC','GD','TOST','FIVE']
print("🚀 Starting CANSLIM Analysis on Sample Stocks")
print("This may take a few minutes to fetch and analyze data...")

results = analyze_multiple_stocks(sample_stocks, period='2y')

# Summary of results
print("\n" + "="*80)
print("CANSLIM ANALYSIS SUMMARY")
print("="*80)

summary_df = pd.DataFrame({
    'Stock': list(results.keys()),
    'Signal': [r['signal'] for r in results.values()],
    'Score': [r['score'] for r in results.values()]
})

summary_df = summary_df.sort_values('Score', ascending=False)
print(summary_df.to_string(index=False))

# Create summary visualization
plt.figure(figsize=(10, 6))
colors = ['red' if s == 'STRONG SELL' or s == 'SELL' else
          'orange' if s == 'HOLD' else
          'lightgreen' if s == 'BUY' else 'green'
          for s in summary_df['Signal']]

bars = plt.bar(summary_df['Stock'], summary_df['Score'], color=colors)
plt.title('CANSLIM Scores Summary')
plt.ylabel('Overall Score (0-100)')
plt.axhline(y=70, color='green', linestyle='--', alpha=0.7, label='Strong Buy Threshold')
plt.axhline(y=60, color='lightgreen', linestyle='--', alpha=0.7, label='Buy Threshold')
plt.axhline(y=40, color='orange', linestyle='--', alpha=0.7, label='Hold/Sell Threshold')
plt.legend()
plt.tight_layout()
plt.show()

# Instructions for custom analysis
print("\n" + "="*80)
print("CUSTOM ANALYSIS INSTRUCTIONS")
print("="*80)
print("To analyze your own stocks, use the following code:")
print()
print("# Single stock analysis")
print("analyzer = CANSLIMAnalyzer('YOUR_SYMBOL', period='2y')")
print("signal, score = analyzer.generate_signals()")
print("analyzer.plot_analysis()")
print()
print("# Multiple stocks analysis")
print("my_stocks = ['STOCK1', 'STOCK2', 'STOCK3']")
print("my_results = analyze_multiple_stocks(my_stocks, period='2y')")
print()
print("Remember:")
print("- CANSLIM is a growth stock methodology")
print("- Scores above 70 indicate strong buy candidates")
print("- Scores below 40 suggest avoiding the stock")
print("- Always consider overall market conditions")
print("- This is for educational purposes - do your own research!")
```
